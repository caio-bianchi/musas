# Experiment 11 — Combined off-nominal conditions

Compare **SD only** and **SD + TD** with localized wind zones and dynamic geofences. This notebook only produces simulation datasets and `replication_metrics.csv`; all analysis and plotting are performed in the separate results notebook.


## Parameters

The numerical configuration is centralized in `_off_nominal_load_common.py`, so all experiments use the same traffic generation, algorithms, dynamics, metrics, and random-seed policy.

| Component | Value |
|---|---:|
| Fleet sizes, $N$ | 5, 10, 15, 20, 25, 30, 35, 40, 45, 50 |
| Replications per case, $K$ | 800 |
| Operating area | $[-200,200] \times [-200,200]$ m |
| Departure interval | 2--80 s |
| Nominal speed | 10 m/s |
| Required / strategic separation | 30 / 40 m |
| Tracking horizon / step | 60 states / 0.5 s |
| Tactical assessment interval | 0.5 s |
| Simulation end | 300 s |

Completed replications already present in `replication_metrics.csv` are skipped, allowing an interrupted sequential execution to continue.


## Scenario

This scenario combines localized wind zones and dynamic geofences while retaining the same paired traffic seeds used by the other load experiments.


```python
from pathlib import Path

from _off_nominal_load_common import (
    K,
    N_VALUES,
    execute_experiment,
    execute_experiment_parallel,
)

SCENARIO = "combined"
EXPERIMENT_LABEL = "11"
OUTPUT_DIRECTORY = Path("outputs") / "output_11"
STRATEGIES = ("sd_only", "sd_td")

print("Scenario:", SCENARIO)
print("Strategies:", STRATEGIES)
print("Fleet sizes:", N_VALUES)
print("Replications per case:", K)
print("Output:", OUTPUT_DIRECTORY)
```

## Sequential execution

The cell below runs only missing replications and appends their metrics to `replication_metrics.csv`. It intentionally performs no plotting.


```python
results = execute_experiment(
    scenario=SCENARIO,
    experiment_label=EXPERIMENT_LABEL,
    output_directory=OUTPUT_DIRECTORY,
    keep_datasets=False,
    strategies=STRATEGIES,
    n_values=N_VALUES,
    replications=K,
)
```

## Optional parallel execution

Run **either** the sequential cell above **or** the parallel cell below. Both modes resume from `replication_metrics.csv` and execute only missing replications. If both cells are run, the second one will normally have no work left.

`MAX_WORKERS` controls how many simulations may run simultaneously. Avoid running the same experiment from two notebook kernels at the same time.


```python
MAX_WORKERS = 6

parallel_results = execute_experiment_parallel(
    scenario=SCENARIO,
    experiment_label=EXPERIMENT_LABEL,
    output_directory=OUTPUT_DIRECTORY,
    keep_datasets=False,
    strategies=STRATEGIES,
    n_values=N_VALUES,
    replications=K,
    max_workers=MAX_WORKERS,
)
```
